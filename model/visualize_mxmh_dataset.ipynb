# การวิเคราะห์ข้อมูล: ดนตรี & สุขภาพจิต

ในโปรเจกต์การวิเคราะห์ข้อมูลนี้ เราจะสำรวจผลการตอบแบบสอบถามเกี่ยวกับดนตรีและสุขภาพจิต เพื่อดูว่ามีนิสัยการฟังเพลงแบบไหนที่เกี่ยวข้องกับสภาวะสุขภาพจิตของผู้ตอบหรือไม่
#### บริบทของชุดข้อมูลและคุณลักษณะ (Features)
ดนตรีบำบัด (Music Therapy หรือ MT) คือ การใช้ดนตรีเพื่อช่วยบรรเทาความเครียด ปรับอารมณ์ และเสริมสร้างสุขภาพจิตโดยรวมให้ดีขึ้น ซึ่งเป็นวิธีการที่ได้รับการยอมรับว่าอ้างอิงจากหลักฐานทางวิทยาศาสตร์ โดยเชื่อว่าดนตรีสามารถกระตุ้นฮอร์โมนแห่งความสุข เช่น ออกซิโทซิน (Oxytocin) ได้

อย่างไรก็ตาม การใช้ดนตรีบำบัดนั้นหลากหลายมาก ทั้งในด้านแนวเพลงและวิธีการ ซึ่งมักจะแตกต่างกันไปในแต่ละองค์กร

ชุดข้อมูล MxMH นี้จัดทำขึ้นเพื่อค้นหาว่า แนวเพลงหรือพฤติกรรมการฟังเพลงของแต่ละบุคคลมีความสัมพันธ์กับสุขภาพจิตที่พวกเขารายงานด้วยตนเองหรือไม่ หากพบว่ามีความเกี่ยวข้อง ข้อมูลนี้อาจนำไปใช้พัฒนาการบำบัดด้วยดนตรี หรืออย่างน้อยก็ช่วยให้เราเข้าใจความเชื่อมโยงระหว่าง “ดนตรี” กับ “จิตใจ” ได้ดียิ่งขึ้น

อายุที่เกิดขึ้นบ่อยที่สุดในชุดข้อมูลคือ 18 ปี // 10 อายุแรกอยู่ระหว่าง 16 ถึง 26 ปี ผู้ตอบส่วนใหญ่เป็นวัยหนุ่มสาว.

สามารถสร้างหมวดหมู่อายุได้ เช่น วัยรุ่น, วัยหนุ่มสาว, วัยผู้ใหญ่ และวัยชรา โดยให้ x ที่เป็นอายุของผู้เข้าร่วม:
- 0 < x < 18 สามารถเป็นวัยรุ่น,
- 18 < x < 25 สามารถเป็นวัยหนุ่มสาว,
- 25 < x < 40 สามารถเป็นผู้ใหญ่,
- 40 < x สามารถเป็นวัยชรา.

```python
df['Age_Category'] = pd.cut(x = df['Age'],
                            bins = [0, 18, 25, 40, np.max(df['Age'])],
                            labels = ['Teenager', 'Young', 'Mature', 'Old'],
                            right = False)

df['Age_Category']
```

```python
age_categories = df.Age_Category.value_counts().index.values
age_category_freq = df.Age_Category.value_counts().values
```

```python
sns.barplot(x = age_categories, y = age_category_freq)

plt.title('Frequencies of Age Categories')
plt.xlabel('Category')
plt.ylabel('Frequency')
plt.show()
```

Bar Chart ที่ได้เเสดงว่ากลุ่มที่ตอยส่วนใหญ่จะเป็นวัยหนุ่มสาว

```python
df['Primary streaming service'].isna().sum()
```

1. เเทนที่ค่า "I do not use a streaming service." เป็น "Not",

```python
df['Primary streaming service'] = df['Primary streaming service'].replace({
    'I do not use a streaming service.' : 'Not',
    'Other streaming service' : 'Other'
})

df['Primary streaming service'].value_counts() 
```

ช่วงเวลาระหว่าง 5 ถึงน้อยกว่า 10 ชั่วโมง (5 ≤ x < 10) เราจะเรียกว่า "ปกติ"

```python
df['Hours_Category'] = pd.cut(x = df['Hours per day'],
                              bins = [0, 5, 10, np.max(df['Hours per day'])],
                              right = False,
                              labels = ['Less', 'Normal', 'More'])
```
